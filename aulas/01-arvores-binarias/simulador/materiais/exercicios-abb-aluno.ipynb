## O código da aula
O mesmo código que roda no simulador. Rode a célula e compare a saída com o console do simulador.

```python
class No:
  def __init__(self, valor):
    self.valor = valor
    self.esquerda = None
    self.direita = None

class ABBCompleta:
  def inserir(self, raiz, valor):
    if raiz is None:
      return No(valor)
    if raiz.valor > valor:
      raiz.esquerda = self.inserir(raiz.esquerda, valor)
    else:
      raiz.direita = self.inserir(raiz.direita, valor)
    return raiz

  def buscar(self, raiz, valor):
    if raiz is None or raiz.valor == valor:
      return raiz
    if raiz.valor > valor:
      return self.buscar(raiz.esquerda, valor)
    return self.buscar(raiz.direita, valor)

  # Travessia Em-Ordem (Esquerda, Raiz, Direita) - Retorna os elementos ordenados
  def em_ordem(self, raiz):
    elementos = []
    if raiz:
      elementos.extend(self.em_ordem(raiz.esquerda))
      elementos.append(raiz.valor)
      elementos.extend(self.em_ordem(raiz.direita))
    return elementos

  # Travessia Pré-Ordem (Raiz, Esquerda, Direita)
  def pre_ordem(self, raiz):
    elementos = []
    if raiz:
      elementos.append(raiz.valor)
      elementos.extend(self.pre_ordem(raiz.esquerda))
      elementos.extend(self.pre_ordem(raiz.direita))
    return elementos

  # Travessia Pós-Ordem (Esquerda, Direita, Raiz)
  def pos_ordem(self, raiz):
    elementos = []
    if raiz:
      elementos.extend(self.pos_ordem(raiz.esquerda))
      elementos.extend(self.pos_ordem(raiz.direita))
      elementos.append(raiz.valor)
    return elementos

# Criando e populando a nova árvore
raiz_completa = None
abb = ABBCompleta()
numeros = [45, 20, 60, 10, 30, 50, 70]

for num in numeros:
  raiz_completa = abb.inserir(raiz_completa, num)

# Exibindo os resultados dos métodos clássicos
print("Travessia Em-Ordem (Ordenado):", abb.em_ordem(raiz_completa))
print("Travessia Pré-Ordem:", abb.pre_ordem(raiz_completa))
print("Travessia Pós-Ordem:", abb.pos_ordem(raiz_completa))

# Teste de busca
valor_teste = 30
resultado_busca = abb.buscar(raiz_completa, valor_teste)
print(f"\nBusca pelo valor {valor_teste}:", "Encontrado!" if resultado_busca else "Não encontrado.")
```

## Ferramentas da aula

```python
# Ferramentas da aula: rode esta célula uma vez (não precisa alterar nada aqui)
import hashlib

ARVORES = {'aula': [45, 20, 60, 10, 30, 50, 70], 'ordenada': [10, 20, 30, 45, 50, 60, 70], 'um_no': [42], 'vazia': [], 'maior': [50, 30, 70, 20, 40, 60, 80, 35, 65]}

def montar(lista):
  raiz = None
  for v in lista:
    raiz = abb.inserir(raiz, v)
  return raiz

_ESPERADO = {"contar_nos": [["aula", [], "7902699be42c8a8e"], ["ordenada", [], "7902699be42c8a8e"], ["um_no", [], "6b86b273ff34fce1"], ["vazia", [], "5feceb66ffc86f38"], ["maior", [], "19581e27de7ced00"]], "altura": [["aula", [], "4e07408562bedb8b"], ["ordenada", [], "7902699be42c8a8e"], ["um_no", [], "6b86b273ff34fce1"], ["vazia", [], "5feceb66ffc86f38"], ["maior", [], "4b227777d4dd1fc6"]], "minimo": [["aula", [], "4a44dc15364204a8"], ["ordenada", [], "4a44dc15364204a8"], ["um_no", [], "73475cb40a568e8d"], ["vazia", [], "dc937b59892604f5"], ["maior", [], "f5ca38f748a1d6ea"]], "maximo": [["aula", [], "ff5a1ae012afa5d4"], ["ordenada", [], "ff5a1ae012afa5d4"], ["um_no", [], "73475cb40a568e8d"], ["vazia", [], "dc937b59892604f5"], ["maior", [], "48449a14a4ff7d79"]], "contar_folhas": [["aula", [], "4b227777d4dd1fc6"], ["ordenada", [], "6b86b273ff34fce1"], ["um_no", [], "6b86b273ff34fce1"], ["vazia", [], "5feceb66ffc86f38"], ["maior", [], "4b227777d4dd1fc6"]], "caminho_busca": [["aula", [30], "fa8f4a97d9d31bc3"], ["aula", [55], "5498d97b9db2b338"], ["ordenada", [70], "b1edd1ea9a4f15b1"], ["vazia", [5], "4f53cda18c2baa0c"], ["maior", [65], "6c67ea6e5aa65ec1"], ["maior", [36], "dd272b03459f0a14"]]}

def _h(v):
  return hashlib.sha256(repr(v).encode()).hexdigest()[:16]

def verificar(funcao):
  """Testa a sua função em várias árvores. Mostra o que ela devolveu, mas não a resposta certa."""
  nome = funcao.__name__
  acertos = 0
  casos = _ESPERADO[nome]
  for arvore, args, esperado in casos:
    descricao = f"árvore {arvore} {ARVORES[arvore]}" + (f", valor {args[0]}" if args else "")
    try:
      obtido = funcao(montar(ARVORES[arvore]), *args)
    except Exception as erro:
      print(f"✖ {descricao}: deu erro → {type(erro).__name__}: {erro}")
      continue
    if _h(obtido) == esperado:
      acertos += 1
      print(f"✔ {descricao}")
    elif obtido is None and nome not in ("minimo", "maximo"):
      print(f"✖ {descricao}: sua função devolveu None. Faltou um return?")
    else:
      print(f"✖ {descricao}: sua função devolveu {obtido!r}, mas não é isso.")
  print()
  if acertos == len(casos):
    print(f"🎉 {nome}: todos os {len(casos)} testes passaram! Agora confira no simulador e anote na ficha.")
  else:
    print(f"{nome}: {acertos} de {len(casos)} testes passaram. Releia as dicas e tente de novo.")

print("Ferramentas prontas. Árvores de teste:", ", ".join(ARVORES))
```
